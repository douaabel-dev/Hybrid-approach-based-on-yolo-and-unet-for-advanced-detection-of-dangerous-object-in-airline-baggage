```python
data_yaml_content = """
path: /kaggle/input/datasets/houdabelhadjadji/my-new-dataset/train_images_without_boxes
train: train/images
val: val/images

nc: 12
names:
  0: Baton
  1: Pliers
  2: Hammer
  3: Powerbank
  4: Scissors
  5: Wrench
  6: Gun
  7: Bullet
  8: Sprayer
  9: HandCuffs
  10: Knife
  11: Lighter
"""

with open("/kaggle/working/data.yaml", "w") as f:
    f.write(data_yaml_content)

print("data.yaml created successfully!")


import os
import numpy as np
from pathlib import Path
from ultralytics import YOLO
from ultralytics import YOLO

model = YOLO("yolo11n.pt")  # or yolov8n.pt depending on your run

model.train(
    data="data.yaml",
    epochs=100,
    imgsz=640,
    batch=16,
    device=0,
    project="runs/train",
    name="exp_augmented",
    
    # ───── AUGMENTATIONS ─────
    hsv_h=0.015,
    hsv_s=0.7,
    hsv_v=0.4,

    degrees=10.0,
    translate=0.1,
    scale=0.5,
    shear=2.0,

    fliplr=0.5,

    mosaic=1.0,
    mixup=0.1,

    save=True
)
```
